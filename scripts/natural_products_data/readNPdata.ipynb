### Combine the data bases

```python
import warnings
from rdkit import Chem
from rdkit.Chem import PandasTools
from rdkit import RDLogger
import pandas as pd

# Suppress warnings
RDLogger.DisableLog('rdApp.*')
warnings.filterwarnings('ignore')


def canonicalize_smiles(smi: str) -> str:
    """Convert SMILES to canonical form."""
    try:
        if pd.isna(smi) or not smi:
            return None
        mol = Chem.MolFromSmiles(str(smi))
        if mol:
            return Chem.MolToSmiles(mol, canonical=True)
    except:
        pass
    return None


# Load databases
print("Loading databases...")
npatlasDF = PandasTools.LoadSDF('NPAtlas_np_data.sdf', smilesName='SMILES', molColName='Molecule')
coconutDF = PandasTools.LoadSDF('coconut_np_data.sdf', smilesName='SMILES', molColName='Molecule')

print(f"NPAtlas: {len(npatlasDF)} molecules")
print(f"COCONUT: {len(coconutDF)} molecules")

# Canonicalize and extract essential columns from NPAtlas
print("\nProcessing NPAtlas...")
npatlas_essential = pd.DataFrame({
    'Canonical_SMILES': npatlasDF['compound_smiles'].apply(canonicalize_smiles),
    'InChIKey': npatlasDF['compound_inchikey'],
    'Name': npatlasDF['compound_name'],
    'Molecular_Formula': npatlasDF['compound_molecular_formula'],
    'Molecular_Weight': npatlasDF['compound_molecular_weight'],
    'Database_ID': npatlasDF['npaid'],
    'Database': 'NPAtlas',
    'Origin': npatlasDF['origin_species']
})

# Canonicalize and extract essential columns from COCONUT
print("Processing COCONUT...")
coconut_essential = pd.DataFrame({
    'Canonical_SMILES': coconutDF['canonical_smiles'].apply(canonicalize_smiles),
    'InChIKey': coconutDF['standard_inchi_key'],
    'Name': coconutDF['name'],
    'Molecular_Formula': coconutDF['molecular_formula'],
    'Molecular_Weight': coconutDF['molecular_weight'],
    'Database_ID': coconutDF['identifier'],
    'Database': 'COCONUT',
    'Origin': coconutDF['organisms']
})

# Combine
print("\nCombining...")
naturalProductDF = pd.concat([npatlas_essential, coconut_essential], ignore_index=True)

# Remove invalid SMILES
naturalProductDF = naturalProductDF.dropna(subset=['Canonical_SMILES'])

# Remove duplicates (keep entry with name if available)
naturalProductDF['has_name'] = naturalProductDF['Name'].notna()
naturalProductDF = naturalProductDF.sort_values(by=['Canonical_SMILES', 'has_name'], ascending=[True, False])
naturalProductDF = naturalProductDF.drop_duplicates(subset=['Canonical_SMILES'], keep='first')
naturalProductDF = naturalProductDF.drop(columns=['has_name']).reset_index(drop=True)

# Summary
print("NATURAL PRODUCTS DATABASE")
print(f"Total unique compounds: {len(naturalProductDF)}")
print(f"\nBy database:")
print(naturalProductDF['Database'].value_counts())
print(f"\nColumns: {naturalProductDF.columns.tolist()}")

# Save
naturalProductDF.to_csv('naturalProductDF.csv', index=False)
naturalProductDF.to_parquet('naturalProductDF.parquet', index=False)
print("\nSaved to naturalProductDF.csv and naturalProductDF.parquet")

naturalProductDF
```

### generate canonical SMILES from the molecule objects

```python
%%time

import warnings
from rdkit import Chem
from rdkit.Chem import PandasTools, rdMolDescriptors
from rdkit.Chem.inchi import MolToInchi, InchiToInchiKey
from rdkit import RDLogger
import pandas as pd

# Suppress warnings
RDLogger.DisableLog('rdApp.*')
warnings.filterwarnings('ignore')


def mol_to_canonical_smiles(mol) -> str:
    """Convert RDKit mol object to canonical SMILES (same as code-2)."""
    try:
        if mol:
            return Chem.MolToSmiles(mol, canonical=True)
    except:
        pass
    return None


def mol_to_inchikey(mol) -> str:
    """Convert RDKit mol object to InChIKey (same as code-2)."""
    try:
        if mol:
            inchi = MolToInchi(mol)
            if inchi:
                return InchiToInchiKey(inchi)
    except:
        pass
    return None


def get_mol_property(mol, property_names: list) -> str:
    """Get first available property from molecule."""
    if not mol:
        return None
    props = mol.GetPropsAsDict()
    for prop_name in property_names:
        if prop_name in props:
            val = props[prop_name]
            if val and str(val).strip():
                return str(val).strip()
    return None


def process_sdf_to_dataframe(sdf_path: str, database_name: str) -> pd.DataFrame:
    """
    Process SDF file and generate canonical SMILES from molecule objects.
    This mirrors the behavior of code-2's load_natural_products_from_sdf().
    """
    print(f"\nProcessing: {sdf_path}")
    print(f"Database: {database_name}")
    
    # Property mappings (same as code-2)
    name_properties = {
        'npatlas': ['compound_name', 'compound_names', 'npaid', 'name', 'ID'],
        'coconut': ['name', 'coconut_id', 'Name', 'ID', 'compound_name'],
    }
    
    id_properties = {
        'npatlas': ['npaid', 'npatlas_id', 'ID', 'id'],
        'coconut': ['identifier', 'coconut_id', 'ID', 'id'],
    }
    
    source_properties = {
        'npatlas': ['origin_species', 'origin_type', 'genus', 'species'],
        'coconut': ['organisms', 'source', 'organism', 'textTaxa'],
    }
    
    db_key = database_name.lower()
    
    # Load using SDMolSupplier (same as code-2)
    supplier = Chem.SDMolSupplier(sdf_path)
    
    if supplier is None:
        raise ValueError(f"Could not load SDF file: {sdf_path}")
    
    records = []
    valid_count = 0
    error_count = 0
    
    for idx, mol in enumerate(supplier):
        if mol is None:
            error_count += 1
            continue
        
        # Generate canonical SMILES from molecule object (CRITICAL - same as code-2)
        canonical_smiles = mol_to_canonical_smiles(mol)
        if not canonical_smiles:
            error_count += 1
            continue
        
        # Generate InChIKey from molecule object (same as code-2)
        inchikey = mol_to_inchikey(mol)
        
        # Get properties
        name = get_mol_property(mol, name_properties.get(db_key, name_properties['npatlas']))
        db_id = get_mol_property(mol, id_properties.get(db_key, id_properties['npatlas']))
        origin = get_mol_property(mol, source_properties.get(db_key, source_properties['npatlas']))
        
        # Calculate molecular properties from mol object
        try:
            formula = rdMolDescriptors.CalcMolFormula(mol)
            mol_weight = rdMolDescriptors.CalcExactMolWt(mol)
        except:
            formula = None
            mol_weight = None
        
        records.append({
            'Canonical_SMILES': canonical_smiles,
            'InChIKey': inchikey,
            'Name': name,
            'Database_ID': db_id,
            'Origin': origin,
            'Molecular_Formula': formula,
            'Molecular_Weight': mol_weight,
            'Database': database_name
        })
        
        valid_count += 1
        
        if (idx + 1) % 50000 == 0:
            print(f"  Processed {idx + 1} molecules...")
    
    print(f"  Valid: {valid_count}, Errors: {error_count}")
    
    return pd.DataFrame(records)


# Process NPAtlas
print("Creating Natural Products Database")

npatlas_df = process_sdf_to_dataframe(
    'NPAtlas_np_data.sdf', 
    'NPAtlas'
)
print(f"NPAtlas: {len(npatlas_df)} molecules")

# Process COCONUT
coconut_df = process_sdf_to_dataframe(
    'coconut_np_data.sdf', 
    'COCONUT'
)
print(f"COCONUT: {len(coconut_df)} molecules")

# Combine
print("\n" + "-" * 60)
print("Combining and deduplicating...")
naturalProductDF = pd.concat([npatlas_df, coconut_df], ignore_index=True)
print(f"Combined: {len(naturalProductDF)} molecules")

# Remove invalid SMILES
naturalProductDF = naturalProductDF.dropna(subset=['Canonical_SMILES'])
print(f"After removing invalid SMILES: {len(naturalProductDF)} molecules")

# Remove duplicates (keep entry with name if available)
naturalProductDF['has_name'] = naturalProductDF['Name'].notna()
naturalProductDF = naturalProductDF.sort_values(
    by=['Canonical_SMILES', 'has_name', 'Database'],
    ascending=[True, False, True]
)
before_dedup = len(naturalProductDF)
naturalProductDF = naturalProductDF.drop_duplicates(subset=['Canonical_SMILES'], keep='first')
naturalProductDF = naturalProductDF.drop(columns=['has_name']).reset_index(drop=True)

print(f"After deduplication: {len(naturalProductDF)} unique molecules")
print(f"Duplicates removed: {before_dedup - len(naturalProductDF)}")

# Summary
print("\n" + "=" * 60)
print("SUMMARY")
print("=" * 60)
print(f"Total unique compounds: {len(naturalProductDF)}")
print(f"\nBy database:")
print(naturalProductDF['Database'].value_counts())

print(f"\nData quality:")
print(f"  With names: {naturalProductDF['Name'].notna().sum()} ({100*naturalProductDF['Name'].notna().mean():.1f}%)")
print(f"  With InChIKey: {naturalProductDF['InChIKey'].notna().sum()} ({100*naturalProductDF['InChIKey'].notna().mean():.1f}%)")

# Save
print("\n" + "-" * 60)
print("Saving...")
naturalProductDF.to_csv('naturalProductDF.csv', index=False)
naturalProductDF.to_parquet('naturalProductDF.parquet', index=False)
print("Saved to naturalProductDF.csv and naturalProductDF.parquet")
```
